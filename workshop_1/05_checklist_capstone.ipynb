### Self-check script (runnable) — verifies artifacts on disk

```python
import os, pathlib, json, subprocess

home = pathlib.Path.home()
todo = home / 'workshop-todo'
checks = {
    'workshop-todo project':        todo.exists(),
    'CLAUDE.md (project)':          (todo / 'CLAUDE.md').exists(),
    'CLAUDE.md (user)':             (home / '.claude' / 'CLAUDE.md').exists(),
    'tests/CLAUDE.md (subdir)':     (todo / 'tests' / 'CLAUDE.md').exists(),
    'hook in settings.json':        (todo / '.claude' / 'settings.json').exists(),
    'ship skill':                   (todo / '.claude' / 'skills' / 'ship' / 'SKILL.md').exists(),
    'custom agent':                 (todo / '.claude' / 'agents').exists(),
    'git repo':                     (todo / '.git').exists(),
}
try:
    out = subprocess.run(['claude', 'mcp', 'list'], capture_output=True, text=True, timeout=30).stdout
    checks['at least one MCP server'] = 'No MCP servers' not in out and out.strip() != ''
except Exception:
    checks['at least one MCP server'] = False

done = sum(checks.values())
for k, v in checks.items():
    print(('✅' if v else '⬜'), k)
print(f'\n{done}/{len(checks)} artifacts found')
```
