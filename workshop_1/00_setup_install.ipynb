## Lab 1b — First launch & sign-in
> ⌨️ **Interactive step — do this in a real terminal**, not in this notebook. Claude Code operates primarily as an interactive Terminal User Interface (TUI) running directly inside your command-line environment; the notebook cells below only run *headless* (`claude -p`) commands.

```bash
mkdir -p ~/workshop-todo && cd ~/workshop-todo
git init
claude
# 1st run: browser opens -> sign in -> paste code back
# 'Do you trust the files in this folder?' -> Yes
```
**First 5 things to type at the `>` prompt:**
| Command | Does |
|---|---|
| `/help` | every slash command, flags, shortcuts |
| `/status` | account, model, working dir, session id |
| `/model` | switch model (Opus / Sonnet / Haiku) for this session |
| `/login` / `/logout` | switch account / sign out |
| `/exit` | quit (also Ctrl+C twice, or Ctrl+D) |

**Interactive basics:** `Enter` send · `Shift+Enter` newline · `Esc` interrupt mid-turn ·
`Shift+Tab` cycles permission mode (default → auto-accept edits → plan mode) ·
`Tab` autocompletes `@files` and `/commands` · `↑` prompt history.
